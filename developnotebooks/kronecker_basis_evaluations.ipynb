```python
from ngo.numericalanalysis.basisfunctions import *
from ngo.numericalanalysis.quadrature import *
import matplotlib.pyplot as plt
```

```python
N = 200
n_t = 2
n_x = 10
n_y = 10
Q_t = 3
Q_x = 99
Q_y = 99

n_elements_t = 1
n_elements_x = 3
n_elements_y = 3

basis_t = BSplineBasis(N=n_t, p=1, C=0, dtype=torch.float64, device='cuda')
basis_x = BSplineBasis(N=n_x, p=3, C=2, dtype=torch.float64, device='cuda')
basis_y = BSplineBasis(N=n_y, p=3, C=2, dtype=torch.float64, device='cuda')

basis = TensorProductBasis([basis_t, basis_x, basis_y])

w_t, t = GaussLegendreQuadrature(Q=Q_t, n_elements=n_elements_t, dtype=torch.float64, device='cuda').compute_quadrature()
w_x, x = GaussLegendreQuadrature(Q=Q_x, n_elements=n_elements_x, dtype=torch.float64, device='cuda').compute_quadrature()
w_y, y = GaussLegendreQuadrature(Q=Q_y, n_elements=n_elements_y, dtype=torch.float64, device='cuda').compute_quadrature()


u_m = torch.rand(N, dtype=torch.float64, device='cuda')
U = u_m.reshape((1,n_t, n_x, n_y))

W = w_t, w_x, w_y
X = t, x, y
```

```python
Phi_t = basis_t.forward(t)
Phi_x = basis_x.forward(x)
Phi_y = basis_y.forward(y)
```
