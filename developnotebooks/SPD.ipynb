```python
import torch
```

```python
A = torch.rand(100,200,200, dtype=torch.float32)*2 - 1
L = torch.tril(A)
diag_idx = torch.arange(200, dtype=torch.long)
L[:, diag_idx, diag_idx] = torch.nn.functional.softplus(L[:, diag_idx, diag_idx])
A = L@L.transpose(-2,-1)
# A = 0.5*(A + A.transpose(-2,-1))  # make it symmetric
A = A + 1e-4*torch.eye(200, dtype=torch.float32)
torch.amin(torch.real(torch.linalg.eigvals(A)), axis=-1)
```
